```python
#Soal 4
print('Soal No 4')
peserta = ['Ijul','Chlea','Reza','Citra','Budi']
for p  in peserta:
    if p == 'Reza':
        continue
    print(p)
```

```python
#Soal 5
print('Soal No 5')
departemen = ['Statbis','Statistika','Bisnis']
angkatan = [2025,2026]
for d in departemen:
    for k in angkatan:
        print(d, f'Angkatan{k}')
```

```python
#Soal 6
print('Soal No 6')
pelamar =[('Budi',3.2,5),('Siti',3.6,3),('Dodi',3.8,1)]
for x, y, z in pelamar:
    if y >= 3.5 and z > 2:
        print(x, "- Diterima")
    else:
        print(x, "- Ditolak")
```

```python
#Soal 7
print('Soal No 7')
karyawan = [('Joko','IT',7000),('Rina','HR',5000),('Toni','Data',5000),('Lia','IT',6500)]
total = 0
for g, h, i in karyawan:
    if h in ['IT','Data']:
        total += i
        print (total)
```

```python
#Soal 8
print('Soal No 8')
penjualan = [("Kopi", 60, False), ("Teh", 30, False), ("Boba", 80, True)]
for p, q, r in penjualan:
    if q > 50 and not r:
        print(p)
```
